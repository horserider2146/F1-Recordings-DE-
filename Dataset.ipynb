## Manual Review — Flagged Clips

Run the cell below. For each clip:
1. Listen to the audio
2. Read the transcript
3. Click the correct emotion label

Progress is saved to `labels.csv` after every click. If you close and reopen the notebook, just re-run the cell — it automatically picks up where you left off.

```python
import pandas as pd
import ipywidgets as widgets
from IPython.display import display, Audio, clear_output
import os

BASE        = r"C:\Users\Ritarshi Roy\OneDrive\Desktop\Projects\F1 Recordings"
LABELS_CSV  = os.path.join(BASE, "annotations", "labels.csv")
TEXT_CSV    = os.path.join(BASE, "annotations", "preprocessed_text.csv")
CLEANED_DIR = os.path.join(BASE, "data", "cleaned")
CLIPS_DIR   = os.path.join(BASE, "data", "clips")

# Reload CSV fresh so resuming works correctly after a break
labels_df = pd.read_csv(LABELS_CSV)
text_df   = pd.read_csv(TEXT_CSV)[["clip_id", "raw_text"]]
df        = labels_df.merge(text_df, on="clip_id", how="left")

flagged = df[df["flagged_for_review"] == True].reset_index(drop=True)
total   = len(flagged)

current = [0]

def get_audio_path(clip_id):
    for folder in [CLEANED_DIR, CLIPS_DIR]:
        path = os.path.join(folder, f"{clip_id}.wav")
        if os.path.exists(path):
            return path
    return None

def save_label(clip_id, label):
    df_save = pd.read_csv(LABELS_CSV)
    df_save.loc[df_save["clip_id"] == clip_id, "final_label"]        = label
    df_save.loc[df_save["clip_id"] == clip_id, "flagged_for_review"] = False
    df_save.to_csv(LABELS_CSV, index=False)

def show_clip(idx):
    clear_output(wait=True)

    if total == 0:
        print("All flagged clips have already been reviewed!")
        return

    if idx >= total:
        print(f"Done! All {total} flagged clips reviewed.")
        return

    row       = flagged.iloc[idx]
    clip_id   = row["clip_id"]
    auto_lbl  = row["final_label"]
    conf      = row["confidence"]
    transcript = row["raw_text"] if pd.notna(row.get("raw_text")) else "(no transcript)"

    progress = widgets.IntProgress(
        value=idx, min=0, max=total,
        description=f"{idx}/{total}",
        style={"bar_color": "#4a90d9"},
        layout=widgets.Layout(width="500px")
    )

    info = widgets.HTML(f"""
        <h3 style="margin-bottom:4px">{clip_id}</h3>
        <p style="margin:4px 0">
            <b>Auto-label:</b> {auto_lbl} &nbsp;|&nbsp;
            <b>Confidence:</b> {conf:.2f}
        </p>
        <p style="margin:4px 0"><b>Transcript:</b></p>
        <div style="background:#f0f4f8;padding:10px;border-radius:6px;
                    max-width:620px;font-size:14px;line-height:1.5">
            {transcript}
        </div>
        <p style="margin-top:10px"><b>Choose the correct label:</b></p>
    """)

    audio_path = get_audio_path(clip_id)
    audio_widget = Audio(filename=audio_path) if audio_path else widgets.Label("(audio not found)")

    btn_calm       = widgets.Button(description="Calm",        button_style="info",    layout=widgets.Layout(width="130px"))
    btn_frustrated = widgets.Button(description="Frustrated",  button_style="warning", layout=widgets.Layout(width="130px"))
    btn_stress     = widgets.Button(description="High Stress", button_style="danger",  layout=widgets.Layout(width="130px"))
    btn_urgent     = widgets.Button(description="Urgent",      button_style="danger",  layout=widgets.Layout(width="130px"))
    btn_skip       = widgets.Button(description="Skip",        button_style="",        layout=widgets.Layout(width="130px"))

    def make_label_handler(label):
        def handler(b):
            save_label(clip_id, label)
            current[0] += 1
            show_clip(current[0])
        return handler

    def on_skip(b):
        current[0] += 1
        show_clip(current[0])

    btn_calm.on_click(make_label_handler("Calm"))
    btn_frustrated.on_click(make_label_handler("Frustrated"))
    btn_stress.on_click(make_label_handler("High Stress"))
    btn_urgent.on_click(make_label_handler("Urgent"))
    btn_skip.on_click(on_skip)

    buttons = widgets.HBox([btn_calm, btn_frustrated, btn_stress, btn_urgent, btn_skip],
                           layout=widgets.Layout(gap="8px", margin="4px 0"))

    display(progress, info, audio_widget, buttons)

show_clip(current[0])
```
